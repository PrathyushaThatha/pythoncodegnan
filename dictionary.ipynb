Dictionary methods
-get (key,default_value:optional):it return the value if key is exits in dictionary otherwise it returns default value

```python
#dictionary keys are unique
details={'age':25,'name':'babu'}
print(details)
```

update (new_dictionary):This method updates the valus if new_dictionary keys already exists in dictionary,otherwise it adds new items to dictionary

```python
details={'age':25,'name':'babu','course':'python','batch':57}
details.update({'batch':58,'course':'java','location':'hyd'})
print(details)
```

keys():this method return the list dictionary 

```python
details={'age':25,'name':'babu','course':'python','batch':57}
keys=details.keys()
print(type(keys))
print(keys)
```

popitem():It return and removes the last item as tuple
pop(key):this method removes the specified key item and return the specified keys value,otherwise keyerror

```python
details={'age':25,'name':'babu','course':'python','batch':57,'location':'hyd'}
item=details.popitem()
print(item)
print(details)
val=details.pop('name')
print(val)
print(details)








```

```python
#membership
#it checks with dictionary keys
details={'age':25,'name':'babu','course':'python','batch':57,'location':'hyd'}
print('babu' in details)
print('name' in details)
```
