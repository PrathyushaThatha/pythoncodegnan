```python
#functions :
#types of functions based on parameters and return value
# 4 .with parameters and with return
# function defination block
def addition(x:int,y:int)->int:
    c=x+y
    return c
#function calling block
a,b=10,20
res=addition(a,b)
print(res)
```

```python
# # 3 .with parameters and without return
#by default function returns none values,when function not returning any value
# function defination block
def addition(x:int,y:int)->int:
    c=x+y
    print(f"This is from function defination block:{c}")
#function calling block
a,b=10,20
res=addition(a,b)
print(f" This is from function calling block:{res}")
```

```python
# # 2.without parameters and with return
# function defination block
def addition()->None:
    x=10
    y=20
    z=x+y
    return z
#function calling block
res=addition()
print(f" This is from function calling block:{res}")
```

```python
# # 1.without parameters and without return
# function defination block
def addition()->None:
    x=10
    y=20
    z=x+y
    print(f"This is from function defination block:{z}")
#function calling block
res=addition()
print(f" This is from function calling block:{res}")
```

```python
#even and odd
def even_or_odd(x:int)->str:
    if x%2==0:
        return "even"
    else:
        return "odd"
#function calling block
n=int(input())
print(even_or_odd(n))
```

```python
#prime numbers
#function defination
def check_prime(n:int)->str:
    if n<2:
        return "not prime"
    for i in range(2,n//2+1):
        if n%i==0:
            return "not prime"
    return "prime"
#function calling
num=int(input())
print(check_prime(num))
        
```
