->types of arguments


```python
#1.positional arguments
#function defination block:here arguments are assigns to parameters as per the sequential of position 
def sub(x:int,y:int)->int:
    z=x-y
    return z
#function calling block
a,b=20,10
print(sub(a,b))
print(sub(b,a))
```

```python
#2 keyword arguments:arguments are explicitly passed with parameter names
#function defination block
def sub(x:int,y:int)->int:
    z=x-y
    return z
#function calling block
a,b=20,10
print(sub(x=a,y=b))
print(sub(y=b,x=a))
```

```python
def add(x:int,y:int)->int:
    z=x+y
    return z
#function calling block
a,b=20,10
print(add(x=a,y=b))
print(add(y=b,x=a))
print(add(5,y=10))
```

```python
#3addition
def add(x:int,y:int,z:int)->int:
    p=x+y+z
    return p
#function calling block
a,b,c=20,10,30
print(add(x=a,y=b,z=c))
print(add(y=b,x=a,z=c))
print(add(z=c,y=b,x=a))
print(add(5,y=10,z=20))
```

```python
#3.Default arguments: the default arguments are optional arguments,if you didnot pass value for default argument,in that case it assigns
#default value otherwise,it assigns new argument value. 
def add(w:int,x:int,y:int=0,z:int=0)->int:
    p=w+x+y+z
    return p
#function calling block
a,b,c,d=map(int,input().split())
print(add(a,b))
print(add(a,b,c))
print(add(a,b,c,d))
```

```python
#default with multiplication
def mul(w:int,x:int,y:int=1,z:int=1)->int:
    p=w*x*y*z
    return p
#function calling block
a,b,c,d=map(int,input().split())
print(mul(a,b))
print(mul(a,b,c))
print(mul(a,b,c,d))
```

```python
#4.variable length position arguments
def add(x,y=0,*args):
    print(type(args))
    print(args)
    s=0
    for val in args:
        s+=val
    return s
print(add(10,20,1,2,3,4,5))
print(add(10,20,1,2,3,4,5,6,7,8,910))



```

```python
#5.variable length keyword argument
def add(x,y=0,*args,**kwargs):
    print(type(kwargs))
    print(kwargs)
    print(kwargs.values())
    print(kwargs.keys())
add(1,2,3,4,5, name="srinu",age=20,course="PFS")
```

```python
#in python every type function calling block shares the arguments refference to parameters
#if the arguments are immutable data type,then behave likes pass by value and if arguments are mutables datatypes(i.e list,set,dict)then
#behave likes pass by refference.
#pass by value
def fun(x):
    print(id(x))
    x+=2
    print("after add2")
    print(id(x))
    print(x)
a=10
fun(a)
print(id(a))
print(a)
```

```python
def fun(x):
    print(id(x))
    x+=2
    print("after add2")
    print(id(x))
    print(x)
a=10
print("before function execution 'a' value :",a)
fun(a)
print("after function execution 'a' value:",a)
print(a)
```

```python
#pass by refference
def add_ele(lst,ele):
    lst.append(ele)
arr=[1,2,3,4]
print("Before function execution arr value:",arr)
add_ele(arr,100)
print("after function  execution arr value:",arr)


```
