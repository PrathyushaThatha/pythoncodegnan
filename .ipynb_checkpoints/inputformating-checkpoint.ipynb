```python
#input formating:
#reading integer from user
num=int(input("enter a number:"))
print(type(num))
print(num)
```

```python
#reading float value from user
val=float(input())
print(type(val))
print(val)
```

```python
#complex reading from user
val=complex(int(input()),int(input()))
print(type(val))
print(val)
```

```python
#bool reading
val=bool(input())
print(type(val))
print(val)
```

```python
#bool reading as integer
val=bool(int(input()))
print(type(val))
print(val)
```

```python
#reading string value from user
s=input()
print(type(s))
print(s)
```

```python
#reading list of integer from users
lst=list(map(int, input().split()))
print(lst)

```

```python
#reading list of string from user
lst=list(map(str, input().split(',')))
print(lst)
```

```python
#approach 2
lst=input().split()
print(lst)
```

```python
#reading tuple of string from user
lst=tuple(map(str, input().split()))
print(lst)
```

```python
##reading set of integer from user
lst=set(map(int, input().split()))
print(lst)
```

```python
#eval():this function evaluates the string type expression
val=eval(input())
print(type(val))
print(val)
```

```python
#1.old style format using comma
name='prathyusha'
age=20
print("My Name is" ,name, "and iam" ,age, "years old")

```

```python
#2 acess specifers
#acess specifiers
#string - %s
#interger - %d
#float - %f
name="srinu"
age=20
#my name is srinu and iam 20 years old
print("My name is %s and iam %d years old"%(name,age))
```

```python
#3 dot format
name="srinu"
age=20
#my name is srinu and iam 20 years old
print("My name is {} and iam {} years old".format(name,age))
print("My name is {} and iam {} years old".format(age,name))
```

```python
#4 f- format
name="srinu"
age=20
#my name is srinu and iam 20 years old
print(f"My name is {name} and iam {age} years old")
print(f"My age is {age} and iam {name} ")
```
