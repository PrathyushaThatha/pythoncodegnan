```python
#by using function return 1 to n
import sys
def fun(n):
    res=[]
    for i in range(1, n+1):
        res.append(i)
    return res
r=fun(10)
print(sys.getsizeof(r))
print(r)
```

```python
import sys
def gen(n):
    res=[]
    for i in range(1, n+1):
        yield i
r=gen(10)
print(r)
n=next(r)
print(sys.getsizeof(n))
print(n)
```

```python
import sys
def gen(n):
    res=[]
    for i in range(1, n+1):
        yield i
r=gen(3)
print(r)
print(next(r))
print(next(r))
print(next(r))
print(next(r))#stop iteration error vasthadhi
```

```python
import sys
def gen(n):
    res=[]
    for i in range(1, n+1):
        yield i
r=gen(3)
print(r)
#to prevent the stop iteration error by using for loop
for val in r:
    print(val)
    
```

```python
#by using function return quebic numbers
import sys
import time
s=time.time()
def fun(n):
    res=[]
    for i in range(1, n+1):
        res.append(i**3)
    return res
r=fun(1000)
print(sys.getsizeof(r))
print(r)
e=time.time()
print(e-s)
```

```python
import time
import sys
s=time.time()
def gen(n):
    for i in range(1,n+1):
        yield i**3
r=gen(1000)
print(r)
for val in r:
    print(val,end="")
print()
e=time.time()
print(e-s)
```
