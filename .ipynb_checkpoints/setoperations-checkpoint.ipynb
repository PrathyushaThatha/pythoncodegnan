```python
set methods
#element adding method
#add(ele):it adds the element to if element not in set,otherwise nothing
s={1,2,3,4,5,4,3}
print(s)
s.add(100)
s.add("code")
print(s)




```

```python
#element removing method 
#remove(item):it removes the item if it exits in set otherwise,'keyerror'
#discard(item):it removes the item if it exits in set otherwise nothing

```

```python
s={1,2,3,4,5}
s.rename(5)
print(s)
#s.remove(100)#keyerror
s.discard(3)
print(s)
s.discard(100)#it
print(s)
```

```python
#set operations
#union(set):It returns the all unique elements from all sets
s1={1,3,5}
s2={4,5,6,7}
s3={1,2,8,9}
union=s1.union(s2,s3)
print(union)
res=s1|s2|s3
print(res)
```

```python
#intersection:tn two set common elements return
s1={1,3,5}
s2={3,5,6,7}
s3={1,5,8,9}
intersection=s1.intersection(s2,s3)
print(intersection)
res=s1&s2&s3
print(res)

```

```python
#difference:it returns the a set elements those are not in anther sets
s1={1,3,5,7}
s2={4,5,6,7}
s3={5,1,2,8,9}
difference=s1.difference(s2,s3)
print(difference)
res=s1-s2-s3
print(res)

```

```python
#symmetric_difference(set):it return the non common elements from all sets
s1={1,3,5,7}
s2={4,5,6,7}
s3={5,1,2,8,9}
sy_diff=s1.symmetric_difference(s2)
print(sy_diff)
res=s1^s2
print(res)

```

```python
#issubset(set)
#issuper(set)
#isdisjoint(set)
s1={3}
s2={4,3,5,6,7}
sub=s1.issubset(s2)
print(sub)
#issuper
s1={1,2,3,4}
s2={4,3,}
super=s1.issuperset(s2)
print(super)
#disjoint
s1={1,2,3,4}
s2={4,3,}
dis=s1.isdisjoint(s2)
print(dis)




```
