```python
#scopes of variables
#1.local scope
#2.enclosing scope
#3.global scope
#4.builtin scope
x=10#global scope variable
def outer():
    y=20#enclosing scope variable
    def inner():
        z=30#local scope variable
        print(z,y,x)#builtin scope function
    inner()
outer()
```

```python
#updating the global scope variable value in local
#can't update global scope variable in local scope dirtly.
#updating the global scope variable value in local scope by using 'global keyword
X=100
def outer():
    global x
    x=x+1
    print('local scope varible value:',x)
outer()
print('global scope variable value:',x)
```

```python
#updating the enclosing scope variable value withlocal scope by using nonlocal
x=10#global scope variable
def outer():
    y=20#enclosing variable
    def inner():
        nonlocal y
        z=30
        y=y+z#local scope variable
        print(z,y,x)#print()-builtin scope function
    inner()
    print('enclosing scope variable value:',y)
outer()
```
