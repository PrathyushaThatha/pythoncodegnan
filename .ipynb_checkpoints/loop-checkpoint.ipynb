```python
#checking weather the number is divisible by3 is fizz and 5 isbuzz and both
nums=list(map(int,input().split()))
res=[]
for val in nums:
    if val%3==0 and val%5==0:
        res.append("Fizzbuzz")
    elif val%3==0:
        res.append("Fizz")
    elif val%5==0:
        res.append("buzz")
    else:
        res.append(val)
print(res)
```

```python
n=int(input())
factors=list()
for i in range(1,n+1):
    if n%i==0:
        factors.append(i)
if len(factors)==2:
    print(f'{n} is prime')
else:
    print(f'{n} is not prime')
    
```

```python
#prime number
n =int(input())
for i in range(2,n//2+1):
    if n%i==0:
        print(f'{n} not prime')
        break
else:
    print(f'{n} is prime')
```

```python
#nestead loop
m=int(input())
for n in range(2,m):
    for i in range(2,n//2+1):
        if n%i==0:
            print(f'{n} not prime')
            break
else:
    print(f'{n} is prime')
    
```
