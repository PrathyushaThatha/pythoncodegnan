```python
#find the 2d and 3d
#2d
#area
#parimeter
#3d
#paremeter
#value 
#surface area
from abc import ABC,abstractmethod
class Shape2D(ABC):
    @abstractmethod
    def area(self):
        pass
    @abstractmethod
    def perimeter(self):
        pass
class Shape3D(ABC):
    @abstractmethod
    def volume(self):
        pass
    @abstractmethod
    def surface_area(self):
        pass
class Rectangle(Shape2D):
    def __init__(self,length,width):
        self.length=length
        self.width=width
    def area(self):
        return f"Rectangle area is {self.length *self.width}"
    def perimeter(self):
        return f"Rectangle perimeter is {2 *(self.length+self.width)}"
class Cube(Shape3D):
    def __init__(self,side):
        self.side=side
    def volume(self):
        return f"Cube volume is {self.side **3}"
    def surface_area(self):
        return f"Cube surface area is {6*(self.side **2)}"
rect=Rectangle(10,5)
print(rect.area())
print(rect.perimeter())
cube=Cube(4)
print(cube.volume())
print(cube.surface_area())
        
```
