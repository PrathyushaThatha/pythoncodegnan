```python
#Recursion
def fun(x):
    if x==0:
        return 0
    fun (x-1)
    print(x)
fun(3)


```

```python
#reverse
def func(x):
    if x==0:
        return 0
    print(x)
    func(x-1)
func(3)
```

```python
#factorial
def factorial(n):
    if n==0 or n==1:
        return 1
    return n*factorial(n-1)
print(factorial(5))
```

```python
#find the sum of digits in a number
#n=456 output=15
def sumofdigits(n):
    if n==0:
        return 0
    return n%10+sumofdigits(n//10)
print(sumofdigits(456))
print(sumofdigits(15346))

    
```

```python
#count digits in a number
def sumofdigits(n):
    if n==0:
        return 0
    return 1+sumofdigits(n//10)
print(sumofdigits(456))
print(sumofdigits(15346))
```

```python
#find the square of a number using lambda
s=lambda x:x**2
print(s(10))
```

```python
#find the square of list of integers using map and lambda
arr=[1,2,3,4,5]
print(list(map(lambda x:x**2,arr)))
```

```python
# check even number form list
arr=[1,2,3,4,5]
print(list(filter(lambda x:x%2==0,arr)))
```

```python
# check number even or odd using lambda function
x=9
res='even' if x%2==0 else 'odd'
print(res)
```

```python
n=int(input())
even_or_odd=lambda x:'even' if x%2==0 else "odd"
print(even_or_odd(n))

```

```python
arr=[1,2,3,4,5]
print(list(map(lambda x:f"{x}-even" if x%2==0 else f"{x}-odd",arr)))
```

```python
#find the frequency of each character
string=input()
count={}
for char in string:
    if char in count:
        count[char]+=1
    else:
        count[char]=1
print(count)
```
