```python
#multi level inheritance
class Person:
    def __init__(self, name, age):  
        self.name = name
        self.age = age
    def details(self):
        return f"Myself {self.name} and I am {self.age} years old"
class Student(Person):
    def __init__(self, name, age, branch):
        super().__init__(name, age)
        self.branch = branch
class Employee(Student):
    def __init__(self, name, age, branch, salary):  
        super().__init__(name, age, branch)
        self.salary = salary
# object creation
e = Employee("Mahi", 20, "Hyd", 50000)
# printing values
print(e.name)
print(e.branch)
print(e.salary)
print(e.details())
```

```python
# Multiple Inheritance Example
class Father:
    def __init__(self, father_name):
        self.father_name = father_name
    def work1(self):
        return "Father working as Engineer"
class Mother:
    def __init__(self, mother_name):
        self.mother_name = mother_name
    def work2(self):
        return "Mother working as Teacher"
# Child class
class Son(Father, Mother):
    def __init__(self, name, age, father_name, mother_name):
        Father.__init__(self, father_name)
        Mother.__init__(self, mother_name)
        self.name = name
        self.age = age
    def Parents_work(self):
        print(self.work1())   
        print(self.work2())
# object creation
s = Son('Srinu', 20, 'Sathya', 'Madhu')
# Method Resolution Order
print(Son.mro())
# Accessing methods
print(s.work1())
print(s.work2())
s.Parents_work()


```

```python
#method overloading
class Math:
    def add(self,a,b):
        return a+b
    def add(self,a,b,c):
        return a+b+c
    def add(self,a,b,c,d):
        return a+b+c+d
m=Math()
print(m.add(1,2,3,4))
#print(m.add(1,2,3))
#print(m.add(1,2))      
```

```python
#method overloading
#Directly python doesnot supports method overloading but,we can achive same functional through the default arguments or arbitary arguments
class Math:
    def add(self,a,b,c=0,d=0):
        return a+b+c+d
m=Math()
print(m.add(1,2,3,4))
print(m.add(1,2,3))
print(m.add(1,2))      
```

```python
# Demonstrating multiple magic methods in one class
class Point:
    def __init__(self, x):
        self.x = x
    def __str__(self):   # String representation
        return f"Point value: {self.x}"
    # Arithmetic magic methods
    def __add__(self, other):
        return Point(self.x + other.x)
    def __sub__(self, other):
        return Point(self.x - other.x)
    def __mul__(self, other):
        return Point(self.x * other.x)
    def __truediv__(self, other):
        return Point(self.x / other.x)
    # Comparison magic methods
    def __eq__(self, other):
        return self.x == other.x
    def __lt__(self, other):
        return self.x < other.x
    def __gt__(self, other):
        return self.x > other.x
# Testing the class
p1 = Point(10)
p2 = Point(5)
print("p1:", p1)
print("p2:", p2)
# Arithmetic operations
print("Addition:", p1 + p2)       
print("Subtraction:", p1 - p2)    
print("Multiplication:", p1 * p2) 
print("Division:", p1 / p2)       
# Comparisons
print("Equal:", p1 == p2)         
print("Less than:", p1 < p2)      
print("Greater than:", p1 > p2)   
```
