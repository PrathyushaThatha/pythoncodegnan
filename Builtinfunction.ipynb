```python
import math
print(math.ceil(4.25))#it returns the upperbound interger value
print(math.floor(4.25))#it returns the lowerbound interger value
print(math.factorial(5))#it returns the factorial of a number
print(math.gcd(10,6,4))#It returns the greatest common devisor
print(math.lcm(10,6,4))#It returns the Least common multiplier
print(math.pow(10,3))# pow(a,b)It returns the Float type of a raise to b
print(math.sqrt(12))#It returns the squareroot value of a number
print(math.log10(4))#To find the log10 values


```

```python
#who will reach the  top(100)in less steps
import random
count=rolls=0
print("Welcome to the who will reach top in less steps game")
while True:
    input("To Trow a dice,please click enter button")
    face_val=random.randint(1,6)
    if count+face_val <=100:
        count+=face_val
    rolls+=1
    if count==100:
        print(f"Your dice value:{face_val}")
        print(f"You reached the top in{rolls}chances")
        break
    print(f"Your dice value:{face_val}")
    print(f"Your at{count}position in {rolls} chances")
    print("*" *50)
    
    
```
