```python
#22/06/2026
units=int(input('enter units:'))
senior=input("enter senior or not:").lower ()=="senior"
bill=0
if 0<units<=100:
    bill=units*1.5
elif 100<units<=200:
    bill=units*2.5
elif 200<units<=500:
    bill=units*4
elif 500< units<=800:
    bill=units*6
else:
    bill=units*6*1.05
if senior:
    bill*=0.9
print(bill)
```

```python
#23/06/2026
#sum of digits in a string
#input:'abc123de456'
#1+2+3+4+5
#output:21
s='abc123de456'
total=0
for ch in s:
    if ch.isdigit():
        total+=int(ch)
print(total)
```

```python
#string palindrome
#input:madam,mom
#output:True
s=input()
if s==s[::-1]:
    print("palindrome")
    print("true")
else:
    print("not palindrome")
```

```python
#number palindrome
#input:12121
output:true
n=input()
if n==n[::-1]:
    print("palindrome")
    print("true")
else:
    print("not palindrome")
```

```python
n = input("Enter a number: ")
reverse = ""
for ch in n:
    reverse = ch + reverse  
if n == reverse:
    print("palindrome")
    print("true")
else:
    print("not palindrome")
    
```
